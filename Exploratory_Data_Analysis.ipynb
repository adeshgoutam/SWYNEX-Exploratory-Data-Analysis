```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
df[['Quantity', 'UnitPrice', 'Revenue']].head()
```

```python
total_revenue = df['Revenue'].sum()

print("Total Revenue:", total_revenue)
```

```python
average_revenue = df['Revenue'].mean()

print("Average Revenue per Transaction:", average_revenue)
```

```python
median_revenue = df['Revenue'].median()

print("Median Revenue:", median_revenue)
```

```python
print("Minimum Revenue:", df['Revenue'].min())
print("Maximum Revenue:", df['Revenue'].max())
```

```python
df['Revenue'].describe()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df['Revenue'], bins=50)

plt.title('Revenue Distribution')
plt.xlabel('Revenue')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Quantity',
    y='Revenue'
)

plt.title('Quantity vs Revenue')
plt.xlabel('Quantity')
plt.ylabel('Revenue')

plt.show()
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df['InvoiceDate'].dtype
```

```python
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
```

```python
df['YearMonth'] = df['InvoiceDate'].dt.to_period('M')
```

```python
df[['InvoiceDate', 'Year', 'Month', 'YearMonth']].head()
```

```python
monthly_sales = df.groupby('YearMonth')['Revenue'].sum().reset_index()

monthly_sales
```

```python
monthly_sales['YearMonth'] = monthly_sales['YearMonth'].astype(str)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales['YearMonth'],
    monthly_sales['Revenue'],
    marker='o'
)

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
highest_month = monthly_sales.loc[
    monthly_sales['Revenue'].idxmax()
]

highest_month
```

```python
lowest_month = monthly_sales.loc[
    monthly_sales['Revenue'].idxmin()
]

lowest_month
```

```python
print("Highest Revenue Month:")
print(highest_month)

print("\nLowest Revenue Month:")
print(lowest_month)
```

```python
top_products_revenue = (
    df.groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_revenue
```

```python
plt.figure(figsize=(12, 6))

top_products_revenue.sort_values().plot(kind='barh')

plt.title('Top 10 Products by Revenue')
plt.xlabel('Revenue')
plt.ylabel('Product')

plt.tight_layout()
plt.show()
```

```python
top_products_quantity = (
    df.groupby('Description')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_quantity
```

```python
plt.figure(figsize=(12, 6))

top_products_quantity.sort_values().plot(kind='barh')

plt.title('Top 10 Products by Quantity Sold')
plt.xlabel('Quantity Sold')
plt.ylabel('Product')

plt.tight_layout()
plt.show()
```

```python
product_summary = (
    df.groupby('Description')
    .agg(
        Total_Quantity=('Quantity', 'sum'),
        Total_Revenue=('Revenue', 'sum')
    )
    .sort_values('Total_Revenue', ascending=False)
)

product_summary.head(10)
```

```python
product_summary.head(10)
```

```python
country_quantity = (
    df.groupby('Country')['Quantity']
    .sum()
    .sort_values(ascending=False)
)

country_quantity.head(10)
```

```python
plt.figure(figsize=(12, 6))

country_quantity.head(10).sort_values().plot(kind='barh')

plt.title('Top 10 Countries by Quantity Sold')
plt.xlabel('Quantity Sold')
plt.ylabel('Country')

plt.tight_layout()
plt.show()
```

```python
country_transactions = (
    df.groupby('Country')['InvoiceNo']
    .nunique()
    .sort_values(ascending=False)
)

country_transactions.head(10)
```

```python
plt.figure(figsize=(12, 6))

country_transactions.head(10).sort_values().plot(kind='barh')

plt.title('Top 10 Countries by Number of Transactions')
plt.xlabel('Number of Transactions')
plt.ylabel('Country')

plt.tight_layout()
plt.show()
```

```python
country_summary = df.groupby('Country').agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Transactions=('InvoiceNo', 'nunique')
).sort_values('Total_Revenue', ascending=False)

country_summary.head(10)
```

```python
unique_customers = df['CustomerID'].nunique()

print("Total Unique Customers:", unique_customers)
```

```python
top_customers_revenue = (
    df.groupby('CustomerID')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customers_revenue
```

```python
plt.figure(figsize=(12, 6))

top_customers_revenue.sort_values().plot(kind='barh')

plt.title('Top 10 Customers by Revenue')
plt.xlabel('Revenue')
plt.ylabel('Customer ID')

plt.tight_layout()
plt.show()
```

```python
top_customers_orders = (
    df.groupby('CustomerID')['InvoiceNo']
    .nunique()
    .sort_values(ascending=False)
    .head(10)
)

top_customers_orders
```

```python
plt.figure(figsize=(12, 6))

top_customers_orders.sort_values().plot(kind='barh')

plt.title('Top 10 Customers by Number of Orders')
plt.xlabel('Number of Orders')
plt.ylabel('Customer ID')

plt.tight_layout()
plt.show()
```

```python
customer_summary = df.groupby('CustomerID').agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Orders=('InvoiceNo', 'nunique'),
    Total_Quantity=('Quantity', 'sum')
).sort_values('Total_Revenue', ascending=False)

customer_summary.head(10)
```

```python
average_customer_revenue = customer_summary['Total_Revenue'].mean()

print("Average Revenue per Customer:", average_customer_revenue)
```

```python
median_customer_revenue = customer_summary['Total_Revenue'].median()

print("Median Revenue per Customer:", median_customer_revenue)
```

```python
negative_quantity = df[df['Quantity'] < 0]

print("Negative Quantity Rows:", len(negative_quantity))
```

```python
print(
    "Percentage:",
    round(len(negative_quantity) / len(df) * 100, 2),
    "%"
)
```

```python
negative_revenue = df[df['Revenue'] < 0]

print("Negative Revenue Rows:", len(negative_revenue))
```

```python
print(
    "Percentage:",
    round(len(negative_revenue) / len(df) * 100, 2),
    "%"
)
```

```python
invalid_price = df[df['UnitPrice'] <= 0]

print("Zero/Negative UnitPrice Rows:", len(invalid_price))
```

```python
missing_customer = df['CustomerID'].isnull().sum()

print("Missing CustomerID:", missing_customer)
```

```python
print(
    "Percentage:",
    round(missing_customer / len(df) * 100, 2),
    "%"
)
```

```python
cancelled = df[
    df['InvoiceNo'].astype(str).str.startswith('C')
]

print("Cancelled Transactions:", len(cancelled))
```

```python
print(
    "Cancellation Percentage:",
    round(len(cancelled) / len(df) * 100, 2),
    "%"
)
```

## Key Business Insights

### 1. Sales Performance
The analysis shows that the business generated significant revenue across the analyzed period. The monthly revenue trend indicates variations in sales performance across different months.

### 2. Product Performance
A small group of products contributes a significant portion of the overall revenue, indicating that certain products are particularly important to business performance.

### 3. Geographic Performance
The country-level analysis shows that some markets contribute substantially more revenue than others, highlighting the importance of focusing on high-performing markets.

### 4. Customer Performance
The customer analysis shows that some customers generate considerably higher revenue than others, indicating the presence of high-value customers.

### 5. Data Quality and Anomalies
The dataset contains unusual observations such as negative quantities, cancelled transactions, missing customer IDs, and extreme values. These should be considered when performing further business analysis.
